## 2.5 Cheat sheet: Changing data

| I want… | Write |
|---|---|
| new column from maths | `df["new"] = df["a"] + df["b"]` |
| True/False column | `df["flag"] = df["a"] > 10` |
| two labels from a condition | `df["c"] = np.where(df["a"] > 10, "hi", "lo")` |
| change some cells | `df.loc[mask, "col"] = value` |
| fill missing with 0 | `df.loc[df["a"].isna(), "a"] = 0` |
| drop columns | `df = df.drop(columns=["a","b"])` |
| rename columns | `df = df.rename(columns={"old": "new"})` |
| sort and keep it | `df = df.sort_values("a")` |
| safe copy to experiment on | `t = df.copy()` |
| subset to change separately | `sub = df[mask].copy()` |

**Rules**
- Methods return a NEW table, so store it: `df = df.method(...)`
- Never `df[...][...] = value` (chained assignment does nothing). Use `df.loc[rows, col] = value`
- Restart & Run All at the end

```python
# B1. Make eng = likes + comments + shares, and eng_rate = eng / followers × 100, rounded to 2. How many eng_rate values are NaN, and why that number? (2.3 recap.)
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

import pandas as pd
import numpy as np
df = pd.read_csv("../data/social_posts.csv")

df["eng"]= df['likes']+df['comments']+df['shares']
df['eng_rate'] = np.round(df['eng'] / df['followers'] * 100, 2)
df['eng_rate'].isna().sum() 
#It’s comments (10) + shares (10) − 1 row missing both = 19.


# B2. Make a column size: "big" if followers ≥ 100,000, otherwise "small". Then:

df['size']= np.where(df['followers']>=100000, "big", "small")

# (a) how many of each? 
(df['size']=='big').sum() #60
(df['size']=='small').sum() #240

# (b) mean likes for big vs small, rounded to 1 (use .loc[mask, "likes"] from 2.4)
np.round(df.loc[df['size']=='big', 'likes'].mean(), 1), np.round(df.loc[df['size']=='small', 'likes'].mean(), 1) 
#(np.float64(2688.8), np.float64(1021.5))

# (c) one sentence: what does this suggest?
#those with followers over 100k have an average of 1500 likes more than those with followers under 100k. Indicates how follower counts increases post response.

# B3. Make a True/False column is_viral (likes > 5000). How many viral posts are there?
df['is_viral']= df['likes']>5000
(df['is_viral']==True).sum() #11 viral posts

# B4. Make clean = df.copy(). In clean only:
clean = df.copy()

# set missing comments to 0 and missing shares to 0 (two .loc lines)
clean.loc[clean["comments"].isna(), "comments"] = 0
clean.loc[clean["shares"].isna(), "shares"] = 0

# recompute eng and eng_rate in clean
clean["eng"]= clean['likes']+clean['comments']+clean['shares']
clean['eng_rate'] = np.round(clean['eng'] / clean['followers'] * 100, 2)

# show: NaN count in clean["eng_rate"], mean clean["eng_rate"] rounded to 2
clean['eng_rate'].isna().sum(), clean["eng_rate"].mean().round(2)

# prove df wasn’t touched: show df["comments"].isna().sum() is still 10
df["comments"].isna().sum()

# B5. In df: rename caption_length → cap_len and drop the date column, keeping both changes. Show df.columns.
df = df.rename(columns={"caption_length": "cap_len"})
df = df.drop(columns=["date"])
df.columns

# B6. Make likes_k = likes in thousands, rounded to 1. Show the first 3 values.
df['likes_k'] = np.round(df['likes']/1000,1)
# df['likes_k']=df.loc[df["likes"]>=1000, "likes"] = df['likes']/1000
df[['likes_k']].head(3)

# B7. Make tiktok = only the TikTok rows (as a separate copy). 
# Add a column bonus = likes × 2 to tiktok. 
#     Show that "bonus" in df.columns is False.

tiktok= df[df["platform"] == "TikTok"].copy()
tiktok['bonus']= tiktok['likes']*2

"bonus" in df.columns, "bonus" in tiktok.columns 
```
