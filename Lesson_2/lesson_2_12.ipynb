When NOT to chain
Chains longer than ~10 steps get hard to read. Split them into a function plus a chain.
If you need to look at the data between steps (exploring), step-by-step cells are fine. Chain the final version.

## 2.12 Cheat sheet: Method chaining

| I want… | Write |
|---|---|
| multi-line chain | wrap in `( ... )`, one `.step()` per line |
| filter in a chain | `.query("a > 5 and b == 'x'")` |
| filter by list | `.query("b in ['x', 'y']")` |
| use a Python variable | `.query("a > @threshold")` |
| add a column in a chain | `.assign(new=lambda d: d["a"] * 2)` |
| column using another new one | same `.assign`, later line uses the earlier name |
| custom step | `.pipe(my_function)` (function returns a DataFrame) |
| debug a chain | comment out later lines |
| reusable cleaning | `def load(path): return (pd.read_csv(path) .assign(...) ...)` |

**Rules**
- No outer brackets → SyntaxError on the next line
- query: `and`/`or` + quotes inside; masks: `&`/`|` + brackets
- lambda d = the table at this point in the chain
- Raw data is never modified → re-runnable, no stale state

```python
# B5. A cleaning function for influencers (uses 2.9 and 2.11)
def load_posts(path):
    return (
        pd.read_csv(path, parse_dates=["date"])
        .assign(
            eng=lambda d: d["likes"] + d["comments"].fillna(0) + d["shares"].fillna(0),
            month=lambda d: d["date"].dt.month,
        )
    )

posts = load_posts("../data/social_posts.csv")

# (a) Write clean_inf(path) that returns one chain:

def clean_inf(path):
    return (
        pd.read_csv(path, na_values=["-", "unknown"])
            
        .assign(followers= lambda d: d['followers_raw'].str.replace(",", "").astype(float), 
                verified_bool=lambda d: d['verified'].str.lower().map({"yes": True, "no": False}),
                rate= lambda d: d['rate_raw'].str.replace(r"\$", "", regex=True).astype(float),
                country= lambda d: d['country'].fillna('Unknown'),
                tier= lambda d: pd.Categorical(d['tier'],
                                               categories=["bronze", "silver", "gold", "platinum"],
                                               ordered=True),
                likes_was_missing= lambda d: d['avg_likes'].isna(),
                avg_likes= lambda d: d['avg_likes'].fillna(d.groupby('tier', observed=True)['avg_likes'].transform("median")))
        .drop(columns= ['followers_raw', 'rate_raw', 'verified'])
            )                    
# read with na_values=["-", "unknown"]
# assign followers (commas removed → float)            
# assign rate ($ removed → float)
# assign verified_bool (lower → map)
# assign country filled with "Unknown"
# assign tier as an ordered category
# assign likes_was_missing flag
# assign avg_likes filled with the tier median (transform). Order matters: the flag must come before the fill.
# drop followers_raw, rate_raw and verified

# (b) inf = clean_inf("../data/influencers.csv")
inf = clean_inf("../data/influencers.csv")
# (c) Show inf.dtypes and inf.isna().sum().
inf.dtypes
inf.isna().sum()
# (d) Median rate per tier (observed=True).
(
    inf
    .groupby("tier", observed=True)["rate"]
    .median()
)
# (e) Re-run proof: call clean_inf again into inf2, and show that inf.equals(inf2) is True.
inf2= clean_inf("../data/influencers.csv")
inf.equals(inf2)
```

```python
# B6. Merge in a chain (uses 2.7)
df = pd.read_csv("../data/social_posts.csv")
brands = pd.read_csv("../data/brands.csv")
df.head(2)
# (a) One chain:
(
    df
    .merge(brands, on='brand', how='left')
    .dropna(subset=['category'])
    .groupby('category')['likes'].mean()
    .round(1)
    .sort_values(ascending=False)
    
)
# df → left merge with brands
# drop rows with no category (dropna(subset=[...]))
# groupby category → mean likes
# round to 1
# sort highest first
```

```python
# B7. Dates in a chain (uses 2.10)

# (a) One chain:
(
    df
    .assign(date= lambda d: pd.to_datetime(d['date']),
            month= lambda d: d['date'].dt.month)
    .groupby('month').size()
)


(
    df
    .assign(date= lambda d: pd.to_datetime(d['date']))
    .set_index('date')
    .resample('MS')['likes']
    .sum()
)     
# df → assign date converted with pd.to_datetime
# assign month = month number
# groupby month → size()
# (b) A second chain: df → convert date → set_index("date") → resample("MS") → likes sum.
```
