```python
import numpy as np
```

```python
a = np.array([5, 12, 18, 3, 25])
a > 10        # [False  True  True False  True]
```

```python
#defining masks
a[a > 10]     # [12 18 25]
#Masks work on any NumPy array, including text, as long as it’s an array and not a list:
```

```python
names  = np.array(["Ali", "Sara", "Omar"])
scores = np.array([55, 82, 91])
names[scores > 60]        # ['Sara' 'Omar']   ← the mask comes from ONE array, applied to ANOTHER
#The two arrays just need to be the same length, matched by position.
```

```python
#Counting with masks

#True counts as 1 and False as 0, so:

(a > 10).sum()     # 3     → HOW MANY items are > 10
(a > 10).mean()    # 0.6   → what PROPORTION are > 10 (60%)
```

```python
#& and, | or, ~ not
a[(a > 4) & (a < 20)]     # [ 5 12 18]   both true
a[(a < 5) | (a > 20)]     # [ 3 25]      either true
a[~(a > 10)]              # [5 3]        NOT > 10
```

Always use brackets around each condition. & is evaluated before >, so without the brackets you get an error or a wrong result.
Never use and/or/not with arrays. They expect a single True/False, and an array holds many. You get:
ValueError: The truth value of an array with more than one element is ambiguous

```python
b = np.array([4, -2, 7, -5])
b[b < 0] = 0       # b is now [4 0 7 0]
```

```python
# np.where(condition, if_true, if_false) builds a new array by choosing between two values. It’s like the if/else at the front of a comprehension:

np.where(a > 10, "high", "low")     # ['low' 'high' 'high' 'low' 'high']
np.where(a > 20, 20, a)             # caps values at 20: [ 5 12 18  3 20]
```

```python
scores = np.array([[72, 85, 90],
                   [64, 70, 58],
                   [88, 92, 95]])

scores[:, 0] > 70              # [ True False  True]   test column 0
scores[scores[:, 0] > 70]      # rows where column 0 > 70:
                               # [[72 85 90]
                               #  [88 92 95]]
```

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

m.sum()          # 21          everything
m.sum(axis=0)    # [5 7 9]     DOWN the rows    → one result PER COLUMN
m.sum(axis=1)    # [ 6 15]     ACROSS columns   → one result PER ROW

#argmax(axis=...)
```

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])            # shape (2, 3)

m + np.array([10, 20, 30])           # shape (3,) → added to EACH row
# [[11 22 33]
#  [14 25 36]]
```

```python
row_totals = m.sum(axis=1, keepdims=True)    # [[ 6]
                                             #  [15]]   shape (2, 1)
m / row_totals                               # each item divided by ITS row's total
# [[0.167 0.333 0.5  ]
#  [0.267 0.333 0.4  ]]

# Per-row maths with keepdims=True: m.sum(axis=1) gives shape (2,), which won’t broadcast against (2, 3). Add keepdims=True to keep it as a column, (2, 1):

#And column-wise maths needs no keepdims, because (3,) already broadcasts across rows:

m - m.mean(axis=0)      # subtract each column's mean from that column
```

```python
import numpy as np
a = np.array([5, 12, 18, 3, 25])

# A1: [F, T, T, F, T]
print(a > 10)


# A2: [12, 18, 25]
print(a[a > 10])

# A3: 3, 0.6
print((a > 10).sum(), (a > 10).mean())

# A4: [5, 12, 18]
print(a[(a > 4) & (a < 20)])

# A5: [3, 25]
print(a[(a < 5) | (a > 20)])

# A6: [5, 3]
print(a[~(a > 10)])

# A7: which error? (own cell) : ValueError: The truth value of an array with more than one element is ambiguous
# a[(a > 10) and (a < 20)]

# A8: [low, high, high, low, high]
print(np.where(a > 10, "high", "low"))

# A9: [5, 7, 9] [[6], [15]]
m = np.array([[1, 2, 3], 
              [4, 5, 6]])
print(m.sum(axis=0), m.sum(axis=1))

# A10: [[11, 22, 33], [14, 25, 36]]
print(m + np.array([10, 20, 30]))

# A11: which error? (own cell): Value error because (1,3) and (1,2) don't aligh because of 3 vs 2
# m + np.array([10, 20])

# A12: [4, 0, 7, 0]
b = np.array([4, -2, 7, -5])
b[b < 0] = 0
print(b)
```

```python
#b1
stores = np.array(["North", "South", "Central", "East"])
days   = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
sales  = np.array([[120, 135, 150, 110, 160, 200, 210],
                   [ 90,  95, 100,  85, 105, 140, 150],
                   [200, 180, 210, 190, 220, 260, 270],
                   [ 60,  70,  65,  55,  80, 100, 110]])
print(
#(a) The weekly total per store. 
np.sum(sales, axis=1),
# (b) The total per day.
np.sum(sales, axis=0),
# (c) The name of the best store. 
stores[np.argmax(np.sum(sales, axis=1))],
# (d) The name of the busiest day.
days[np.argmax(np.sum(sales, axis=0))]
)
```

```python
#b2
above_150= (sales>150).sum()
prop_150= (sales>150).mean()

above_150, prop_150
```

```python
#b3
sales[:, :5]
#(a) Each store’s weekday average (Mon–Fri) and weekend average (Sat–Sun). Hint: slice the columns, then use axis.
weekday = np.mean(sales[:, :5], axis=1)  
weekend= np.mean(sales[:, 5:], axis=1) 
# (b) The ratio of weekend to weekday average for each store, rounded to 2 decimals.
ratio= np.round(weekend/weekday, 2)
# (c) The names of the stores whose weekend average is at least 50% higher (ratio ≥ 1.5).
stores[ratio>=1.5]
```

```python
#b4
print(
    stores[np.sum(sales, axis=1)>1000],
sales[np.sum(sales, axis=1)>1000]
)
```

```python
#b5
#a) 
sales_copy1= sales.copy()
sales_copy1[sales_copy1>250]= 250

sales_copy2= np.where(sales>250, 250, sales )
print(sales_copy1, sales, sales_copy2, sales, sep="\n")
```

```python
#b6
# Share of the week. For each store, each day’s sales as a percentage of that store’s weekly total, rounded to 1 decimal. Check: each row should add up to (about) 100. Hint: keepdims=True.

weekly_total= sales.sum(axis=1, keepdims=True)
share_percentage=np.round((sales/weekly_total)*100, 1)
print(share_percentage, share_percentage.sum(axis=1, keepdims=True), sep="\n")
```

```python
#b7
store_totals_day= np.mean(sales, axis=0)
subtracted= sales-store_totals_day
subtracted, subtracted.mean(axis=0, keepdims=True)
```
