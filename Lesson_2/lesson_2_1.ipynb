#With lists, a slice is a new copy. With arrays, a slice is a view, a window onto the same data:

a = np.array([1, 2, 3, 4])
s = a[:2]          # a VIEW of a's first two items
s[0] = 99
a                  # [99  2  3  4]   ← the original changed!

# to get copy we use .copy()
s = a[:2].copy()
s[0] = 99          # a is untouched

```python
import numpy as np

# A1: [1,2,3,1,2,3] [2,4,6]
print([1, 2, 3] * 2)
print(np.array([1, 2, 3]) * 2)

# A2: what dtype?: float
print(np.array([1, 2.5, 3]).dtype)

# A3: what happens to the numbers?: They're all str
print(np.array([1, "2", 3]))

# A4: [0,3,6,9]
print(np.arange(0, 10, 3))

# A5: (2,3), 2, 6
m = np.array([[1, 2, 3], 
              [4, 5, 6]])
print(m.shape, m.ndim, m.size)

# A6: 6, [1, 2, 3], [2,5]
print(m[1, 2], m[0], m[:, 1])

# A7: 21, 21/6= 3.5
print(m.sum(), m.mean())

# A8: [11, 22, 33] [10,10,10]
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
print(a + b, a / b)

# A9: the view trap: [1,2] [99, 2, 3, 4]
x = np.array([1, 2, 3, 4])
s = x[:2]
s[0] = 99
print(x)

# A10: [[0,1,2], [3,4,5]] ValueError because elements more for given division
print(np.arange(6).reshape(2, 3))

# A11: which error type? (own cell): ValueError? would work if it was [1,2,3] in 2nd array
# np.array([1, 2, 3]) + np.array([1, 2])
```

## 2D indexing: `[ROWS, COLUMNS]`

The comma separates rows (before) from columns (after). A `:` alone means "all".

| Code | Read it as | Result |
|---|---|---|
| `scores[:2]` | rows up to 2 (all columns) | rows 0–1: `[[72 85 90], [64 70 58]]` |
| `scores[:, 2]` | all rows, column 2 | the third column: `[90 58 95 70]` |
| `scores[2]` | row 2 | `[88 92 95]` |
| `scores[2, :]` | row 2, all columns | same thing: `[88 92 95]` |
| `scores[:2, 2]` | rows 0–1, column 2 | `[90 58]` |
| `scores[2, 1]` | row 2, column 1 | `92` |
