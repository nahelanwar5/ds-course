Rows keep their original label when they move. Analogy: the label is a name tag that the row keeps wearing, and the position is the seat it’s sitting in right now.

## 2.4 Cheat sheet: Selection

| I want… | Write |
|---|---|
| one column (Series) | `df["a"]` |
| some columns (DataFrame) | `df[["a","b"]]` |
| a row by **label** | `df.loc[5]` |
| a row by **position** | `df.iloc[5]` |
| one cell | `df.loc[5, "a"]` / `df.iloc[5, 2]` |
| last row | `df.iloc[-1]` |
| slice by label (**end INCLUDED**) | `df.loc[2:4]` → 2, 3, 4 |
| slice by position (end excluded) | `df.iloc[2:4]` → 2, 3 |
| rows meeting a condition | `df[df["a"] > 10]` |
| condition + columns | `df.loc[df["a"] > 10, ["a","b"]]` |
| and | `(c1) & (c2)` |
| or | `(c1) \| (c2)` |
| not | `~c1` |
| matches one of many | `df["a"].isin(["x","y"])` |
| missing / not missing | `df["a"].isna()` / `df["a"].notna()` |
| count rows matching | `(condition).sum()` |

**Rules**
- `.loc` = labels (name tags), `.iloc` = positions (seats)
- After sorting/filtering, labels ≠ positions. Use `.iloc[0]` for "top row"
- Brackets around every condition when combining
- Use `&` `|` `~`, never `and` `or` `not` (those give "truth value is ambiguous")
