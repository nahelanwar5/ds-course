## 2.10 Cheat sheet: Dates

| I want… | Write |
|---|---|
| text → datetime | `pd.to_datetime(df["d"])` |
| convert while reading | `pd.read_csv(..., parse_dates=["d"])` |
| UK format day/month/year | `pd.to_datetime(s, format="%d/%m/%Y")` |
| bad values → NaT | `errors="coerce"` |
| year / month / day | `.dt.year` / `.dt.month` / `.dt.day` |
| names | `.dt.month_name()` / `.dt.day_name()` |
| weekday number (Mon = 0) | `.dt.dayofweek` |
| weekend flag | `.dt.dayofweek >= 5` |
| filter after a date | `df[df["d"] >= "2026-06-01"]` |
| filter a range | `df["d"].between("2026-03-01", "2026-03-31")` |
| gap between two dates | `(d2 - d1).days` |
| gaps for a column | `(df["d"] - df["d"].min()).dt.days` |
| per week / per month | `df.set_index("d").resample("W")["v"].sum()` (`"MS"` for month) |

**Rules**
- Convert first: date tools only work on datetime, not text
- Default reads 03/02 as March 2 (US). Use format= for UK dates
- Column of dates → `.dt` (like `.str` for text)
- resample needs the date as the index

```python
d = pd.Series(["2026-01-05", "2026-01-10", "2026-02-01"])
dt = pd.to_datetime(d)
```

```python
d.dtype #str
dt.dtype #datetime64

dt.dt.month.tolist() #[1,1,2]

dt.dt.day_name().tolist() #[monday, saturday, sunday]

dt.dt.dayofweek.tolist() #[0, 5, 6]

(dt.dt.dayofweek >= 5).sum() # 2 

(dt.iloc[1] - dt.iloc[0]).days #5

(dt > "2026-01-07").sum() #2

dt.max() - dt.min() #27

pd.to_datetime("03/02/2026").month #3
pd.to_datetime("03/02/2026", format="%d/%m/%Y").month #2

pd.to_datetime(pd.Series(["2026-01-05", "oops"]), errors="coerce").isna().sum() #1
```

```python
# B1. Convert

# (a) Show the dtype of date before.
df['date'].dtype
# (b) Convert date with pd.to_datetime.
df['date']=pd.to_datetime(df['date'])
# (c) Show the dtype after.
df['date'].dtype
# (d) First and last date (.min(), .max()).
df['date'].min(), df['date'].max()
# (e) How many days does the data cover? Give a plain number.
(df['date'].max() - df['date'].min()).days
```

```python
# B2. Month

# (a) Make month = month number.
df['month']= df['date'].dt.month
# (b) Posts per month (.size()).
df.groupby('month').size()
# (c) Mean likes per month, rounded to 1.
df.groupby('month')['likes'].mean().round(1)
# (d) Comment: which month had the most posts? Which had the highest mean likes?
#March, January
```

```python
# B3. Weekday

# (a) Make weekday = day name.
df['weekday']= df['date'].dt.day_name()
# (b) Mean likes per weekday, rounded to 1, sorted highest first.
df.groupby('weekday')['likes'].mean().round(1).sort_values(ascending=False)
# (c) Comment: best and worst day to post?
#best: Wed, worst: Sat
```

```python
# B4. Weekend

# (a) Make is_weekend (True for Sat/Sun).
df['is_weekend']= df['date'].dt.dayofweek >= 5
# (b) How many weekend posts?
df['is_weekend'].sum()
# (c) Mean likes, weekend vs weekday, rounded to 1.
df.groupby('is_weekend')['likes'].mean().round(1)
# (d) Comment: does weekend posting help? (Compare with B3: is it the same story for both weekend days?)
#On the surface it doesn't help-- mean likes lower on weekend. B3 indicated same--Sat was the worst day for posting. However, Sunday is better than most weekdays, meaning weekend avg is dragged down by saturday. 
```

```python
# B5. Filtering

# (a) How many posts are in March? Do it two ways: with & and >= / <, and with .between.
((df['date']>='2026-03-01') & (df['date']<'2026-04-01')).sum()
df[df['date'].between('2026-03-01', '2026-03-31')]
# (b) How many posts are from 1 June onwards?
(df['date']>='2026-06-01').sum()
```

```python
# B6. Days since start

# (a) Make days_since_start = days since the first post (a plain number).
df['days_since_start']= (df['date']-df['date'].min()).dt.days
# (b) What’s its max?
df['days_since_start'].max()
# (c) How many days after the start was the viral post P0019? (Use .loc with a mask, 2.4.)
df.loc[df['post_id']=='P0019', ['days_since_start']]
```

```python
# B7. Weekly posts (resample)

# (a) Posts per week with resample("W") and post_id count.
weekly= df.set_index("date").resample("W")["post_id"].count()
weekly
# (b) How many weeks are there?
len(weekly)
weekly.count()
# weekly.count() works, but len(weekly) is clearer. count skips NaN, so it could undercount weeks with missing values.

# (c) Which week had the most posts? Use idxmax() (2.6).
weekly.idxmax()
# (d) How many posts were in that week?
weekly['2026-03-15 00:00:00']
# (e) Comment: the first week has far fewer posts. Why? (Hint: check what day 1 Jan 2026 was, and what date the first week’s label is.)
#only 4 days since 1st Jan so doesn't count the full 7 days-only thurs till sunday
```

```python
# B8. Monthly likes (resample)

# (a) Total likes per month with resample("MS").
monthly= df.set_index("date").resample("MS")["likes"].sum()
monthly

# (b) Comment: compare to B2. Why do totals and means rank months differently? (Hint: B2(b).)
#march has more ports overall so more likes. january has more likes/no. of posts which means better stats
```
