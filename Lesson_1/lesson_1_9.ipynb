```python
try:
    n = int("abc")          # ← try this
except ValueError:
    n = None                # ← if THAT specific error happens, do this instead
print(n)                    # None, no crash

#Python runs the try block. If the named error happens, it jumps to except. If it doesn't happen, except is skipped.
```

## Common Python Errors

| Error | What it means | Example that causes it |
|---|---|---|
| `SyntaxError` | Python can't read the code (missing `:`, bracket or quote) | `def f(x)` with no colon |
| `IndentationError` | Spaces at the start of a line are wrong | a function body that isn't indented |
| `NameError` | That variable doesn't exist (typo, or the cell wasn't run) | `print(totl)` |
| `TypeError` | Wrong **type** for this operation | `"5" + 3`, or `None + None` |
| `ValueError` | Right type, but a bad **value** | `int("abc")` |
| `IndexError` | That position doesn't exist in the list | `["x"][1]` |
| `KeyError` | That key doesn't exist in the dict | `{"ali": 80}["Ali"]` |
| `AttributeError` | That type doesn't have this method | `[1, 2].items()` |
| `ZeroDivisionError` | You divided by zero | `10 / 0` |

1) Is there a dot .something on the broken line? → probably an AttributeError.
2) Would a different value of the same type fix it? Yes → ValueError. No → TypeError.

**How to read any traceback:**
1. Read the **last line** first: error type + message
2. Find the **lowest `---->` arrow**: the exact line that broke
3. Ask: what value did that line expect, and what did it get?

```python
def safe_float(s):
    try:
        return float(s)
    except ValueError:
        return None

[safe_float(v) for v in ["3.5", "N/A", "10"]]    # [3.5, None, 10.0]
```

```python
#B1
def get_price(item, prices):
    try:
        return prices[item]
    except KeyError as e:
        return 0

def cart_total(cart, prices):
    return sum(get_price(i, prices) for i in cart)

prices = {"pen": 20, "book": 350}

cart_total(["pen", "bag"], prices)
```

```python
#B2
def safe_int(s):
    try:
        s=s.replace(",", "")
        return int(s)
    except ValueError:
        return None

#B3

raw = ["50000", "N/A", "62,000", "", "71000", "unknown"]

safes= [safe_int(salary) for salary in raw]
valid= [safe for safe in safes if safe is not None]
invalid_count= len([invalid for invalid in safes if invalid is None])
valid, invalid_count
```

```python
#B5
def grade(score):  
        if score > 100 or score < 0:
            raise ValueError(f"{score} is an invalid score (must be 0-100)")
        if score>=80:
            return "A"
        if score>=65:
            return "B"
        if score>=50:
            return "C"
        return "F"
    
scores = [91, 65, 49, 80, 50, 120]

for score in scores:
    try: 
        print(grade(score))
    except ValueError as e:
        print("Skipped:", e)
        
```

```python
##B6
# (a): TypeError-- the numbers in list are not in int type, they're string 
total = 0
for n in ["1", "2", "3"]:
    total += int(n)

# (b): KeyError-- dicts are case sensitive
scores = {"ali": 80, "sara": 90}
print(scores["Ali".lower()])

# # (c): SyntaxError-- no colon after function definition
def mean(nums):
    return sum(nums) / len(nums)

# # (d): IndexError-- adding +1 introduces an index beyong the list's range
names = ["a", "b", "c"]
for i in range(len(names)):
    print(names[i])
```

```python
#B7
nums = [4, 8, 12]
try:
    avg = sum(nums) / len(nums) #numz typo -- NameError
except ZeroDivisionError:
    avg = None
print(avg)     # prints 0... but the real average is 8!
```
