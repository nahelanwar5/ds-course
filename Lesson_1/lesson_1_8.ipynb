When NOT to use a comprehension

1) The logic needs more than one if/else. Use a normal loop, or put the logic in a function (section 6).
2) You're printing or changing something else (side effects). Use a loop. [print(x) for x in s] is bad style.
Rule of thumb: if you can't read it out loud in one breath, write a loop instead.

```python
stock = {"pens": 40, "books": 5, "bags": 12}

# Loop over KEYS only (the default)
[k for k in stock]                          # ['pens', 'books', 'bags']

# Loop over VALUES only
[v for v in stock.values()]                 # [40, 5, 12]

# Loop over BOTH, with .items() and unpacking
[f"{k}={v}" for k, v in stock.items()]      # ['pens=40', 'books=5', 'bags=12']

# Transform every value
{k: v * 2 for k, v in stock.items()}        # {'pens': 80, 'books': 10, 'bags': 24}

# Transform every key
{k.upper(): v for k, v in stock.items()}    # {'PENS': 40, 'BOOKS': 5, 'BAGS': 12}

# Filter by value (if at the END)
{k: v for k, v in stock.items() if v < 20}  # {'books': 5, 'bags': 12}
```

```python
orders = [("tea", 3), ("coffee", 5), ("juice", 2)]

[name for name, qty in orders]              # ['tea', 'coffee', 'juice']
[qty for name, qty in orders if qty > 2]    # [3, 5]
sum(qty for name, qty in orders)            # 10

# A list of tuples turned into a dict
{name: qty for name, qty in orders}         # {'tea': 3, 'coffee': 5, 'juice': 2}
```

```python
#B5
sales = [("Lahore", 500), ("Karachi", 300), ("Lahore", 200), ("Islamabad", 700), ("Karachi", 100)]

def group_sum(pairs):
    dic={}
    for key, amount in pairs:
        dic[key]=dic.get(key, 0) + amount
    return dic

def filter_over(dic, threshold=500):
    return {city: sale for city,sale in dic.items() if sale>=threshold}
    
filter_over(group_sum(sales))
```
