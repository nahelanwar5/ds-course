```python
#B1, B2
from utils import safe_divide
from utils import grades
class Student:
    def __init__(self, name, scores=None):
        self.name= name
        self.scores= list(scores) if scores else []
    def add_score(self, score):
        self.scores.append(score)
    def average(self):
        
        self.avg= safe_divide(sum(self.scores),len(self.scores))
        return self.avg
    def grade(self):
        
        avg = self.average()
        if avg is None:
            return None
        return grades(avg)
#B2
    
    def __repr__(self):
        return f"Student(name='{self.name}', score={self.scores})"

ali = Student("Ali")
ali.add_score(80)
ali.add_score(64)
print(ali.average(), ali.grade())    # 72.0 B
print(Student("Zara").average())     # None

#B3
data = {"Ali": [80, 64], "Sara": [91, 88, 95], "Omar": [45, 60], "Hina": [70]}

students=[Student(name, scores) for name, scores in data.items()]
above_avg= [s.name for s in students if s.average()>=70]
sorted_by_score= sorted(students, key=Student.average, reverse=True)
sorted_averages= sorted([s.average() for s in students], reverse=True)
print(sorted_by_score, sorted_averages)
```

```python
#B4
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner=owner
        self.balance=balance
        self.history=[["balance on opening", balance]]
        
    def deposit(self, amount):
            if amount<=0:
                raise ValueError(f"Deposit amount {amount} must be positive")
            else:
                self.history.append(["deposit", amount])
                self.balance+=amount
    
    def withdrawal(self, amount):
            if amount>self.balance:
                raise ValueError(f"balance {self.balance} too low for {amount}")
            else:
                self.history.append(["withdrawal", amount])
                self.balance-=amount


nahel= BankAccount("Nahel", 25000)
nahel.deposit(20000)
nahel.withdrawal(2000)
nahel.balance, nahel.history
# nahel.withdrawal(45000)
nahel.balance, nahel.history
```

```python
#B5
from statistics import mean
class MeanPredictor:
    # fit(self, values): store the mean of values as self.mean_
    # predict(self, n): return a list of n copies of self.mean_
    # predict BEFORE fit → raise ValueError("call fit() first")
    def __init__(self):
        self.mean_ = None
    def fit(self, values):
        self.mean_ = mean(values)
        return self #predict has to run on what fit returned so if there is no 'return self' it runs on nothing
    def predict(self, n):
        if self.mean_ is None:
            raise ValueError("call fit() first")
            
        copies= [self.mean_ for _ in range(n)]
        return copies
        
m = MeanPredictor()
m.fit([10, 20, 30]).predict(3)
# print(m.mean_, m.predict(3))  
# n= MeanPredictor()
# print(n.predict(5)) # 20.0 [20.0, 20.0, 20.0]
```

```python
#check
class Wallet:
    def __init__(self, owner):
        self.owner=owner
        self.amounts=[]
    def add (self, amount):
        self.amounts.append(amount)
    def total(self):
        return sum(self.amounts)
    def largest(self):
        if self.amounts:
            return max(self.amounts)
        return None

n= Wallet("nahel")
n.add(300)
n.add(500)
n.add(1200)

m=Wallet("Aleezae")


print(n.total(), n.largest(), m.largest())
```
