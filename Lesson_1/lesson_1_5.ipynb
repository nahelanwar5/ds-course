person = {"name": "Nahel", "age": 26, "city": "Lahore"}

person["name"]          # "Nahel"
person["age"] = 27      # update an existing key
person["field"] = "DS"  # add a new key
len(person)             # 4

#Keys are usually strings, but can be any unchangeable type (numbers, tuples). Lists can't be keys.

person["salary"]            # ✗ KeyError: the key doesn't exist
person.get("salary")        # None, no error
person.get("salary", 0)     # 0, returns your default if missing
"salary" in person          # False: `in` checks KEYS, not values

del person["field"]         # removes the key
age = person.pop("age")     # removes it AND returns the value

person.keys()     # dict_keys(['name', 'city'])
person.values()   # dict_values(['Nahel', 'Lahore'])
person.items()    # dict_items([('name', 'Nahel'), ('city', 'Lahore')])  pairs as tuples
list(person.keys())   # convert to a normal list if you need indexing

a = {"Ali", "Sara", "Omar"}     # people who bought in January
b = {"Sara", "Omar", "Hina"}    # people who bought in February

a & b    # {'Sara', 'Omar'}               intersection: both months
a | b    # {'Ali', 'Sara', 'Omar', 'Hina'} union: either month
a - b    # {'Ali'}                         difference: January only

Ordered items, may change	list [ ]
Fixed group of values that shouldn't change	tuple ( )
Look up values by name	dict {key: value}
Unique items, fast membership checks	set { }
