```python
#B4

words = "the cat and the hat and the bat"

def count_words(text):
    counts={}
    word_list= text.split()
    for word in word_list:
        counts[word]= counts.get(word, 0)+1
    return counts

def most_common(text):
    counts= count_words(text)
    maxim= max(counts, key=counts.get)
    return maxim
    
count_words(words), most_common(words)

```

```python
#B7

def group_sum(pairs):
    d={}
    for key, amount in pairs:
        d[key]=d.get(key, 0) + amount
    return d
        
def filter_over(d, threshold=500):
    above_threshold={}
    for key, amount in d.items():
        if amount>=threshold:
            above_threshold[key]=amount
    return above_threshold
    
sales = [("Lahore", 500), ("Karachi", 300), ("Lahore", 200), ("Islamabad", 700), ("Karachi", 100)]
expenses = [("food", 20), ("rent", 500), ("food", 15), ("travel", 60)]
totals= group_sum(expenses) #did for sales also
totals
totals, filter_over(totals, threshold= 50)
# print(group_sum([("a", 1)]))
# print(group_sum([("b", 2)]))
```

```python
#last

def count_vowels(text):
    found= set()
    count= 0
    for letter in len(set(text.lower()):    
         if letter in set("aeiou"):
            found.add(letter)
            count+=1
    return count, len(found)

total, unique= count_vowels("Data Science")
total ,unique        

```
