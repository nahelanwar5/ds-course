```python
## UNLIKE STRINGS, LISTS CAN BE CHANGED BY METHODS
scores = [88, 92, 75, 64, 99]

scores[0] = 90          # ✓ works (strings would raise an error)
scores.append(70)       # adds to the end        → [90, 92, 75, 64, 99, 70]
scores.insert(1, 50)    # inserts at position 1  → [90, 50, 92, ...]
scores.remove(64)       # removes the first 64 found
last = scores.pop()     # removes and RETURNS the last item
scores.extend([1, 2])   # adds several items at once
```

```python
# most list methods change the list and return None:
scores = scores.append(5)   # ✗ scores is now None!
scores.append(5)            # ✓ just call it
```

```python
#SORTING

nums = [3, 1, 2]

sorted(nums)     # [1, 2, 3], returns a NEW list; nums is unchanged
nums.sort()      # sorts nums itself and returns None
nums.sort(reverse=True)   # descending order
```

```python
a = [1, 2, 3]
b = a          # b is NOT a copy: it's a second name for the SAME list
b.append(4)
print(a)       # [1, 2, 3, 4], so a changed too!

c = a.copy()   # a real, separate copy
```

```python
#TUPLES CAN'T BE CHANGED
point = (3, 5)        # round brackets
point[0]              # 3, indexing works
point[0] = 9          # ✗ TypeError: tuples can't be changed

x, y = point          # "unpacking": x = 3, y = 5
a, b = b, a           # swap two variables in one line
```

```python
# A4: [3, 1, 2], None 
#.sort() sorts x in place (so x is changed), and it returns None. y catches that return value, so y is None, while x is sorted.
x = [3, 1, 2]
y = x.sort()
print(x, y)
```
