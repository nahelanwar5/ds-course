```python
#check

import random #put imports above always
from collections import Counter

def dice_sums(n, seed):
    
    random.seed(seed)
    sums = [random.randint(1, 6) + random.randint(1, 6) for _ in range(n)]
    counts=Counter(sums)
    sorted_dict = {k: counts[k] for k in sorted(counts, key=counts.get, reverse=True)}
    return sorted_dict

print(dice_sums(1000, 1))
        
```
