```python
#t7
def shipping_cost(weight_kg, express=False):

    express_factor= 2 if express else 1
    
    if weight_kg <= 1:
        return 200*express_factor
    if weight_kg <= 5:
        return 500*express_factor
    cost= 500+(weight_kg-5)*100
    return cost*express_factor


print(shipping_cost(1), shipping_cost(5), shipping_cost(7), shipping_cost(7, express=True))
```

```python
#t12
class Inventory:
    def __init__(self):
        self.dic={}
    def add(self, name, qty):
        if qty<=0:
            raise ValueError("Quantity must exist")
        self.dic[name]= self.dic.get(name, 0)+qty
    def remove(self, name, qty):
        stock = self.dic.get(name, 0)         
        if stock < qty:                        
            raise ValueError(f"only {stock} {name} in stock, cannot remove {qty}")
        self.dic[name] -= qty
    def total_items(self):
            return sum(self.dic.values())
    def most_stocked(self):
        if not self.dic:
            return None
        return max(self.dic, key=self.dic.get)
    def __repr__(self):
        # Inventory(3 items, 42 units)
        return f"Inventory({len(self.dic)} items, {sum(self.dic.values())} units)"

inv = Inventory()
inv.add("Apple", 10)
print(inv.dic)                      # add
inv.remove("Apple", 5)
print(inv.dic)                      # remove
print(inv.total_items())        # total_items
print(inv.most_stocked())       # most_stocked

inv.add("Apple", 10)
inv.add( "Orange", 5)
print(inv.dic)
inv.remove("Apple", 3)
print(inv.dic)
print(inv.total_items())
print(inv.most_stocked())
try:
    inv.remove("cup", 1)
except ValueError as e:
    print("Failed:", e)
inv.remove("cup", 3)
print(inv)
```

```python
#MINI PROJECT
import csv
import utils as funcs
import json

rows = [
    ["date", "category", "amount"],
    ["2026-10-01", "food", "1,250"],
    ["2026-10-01", "transport", "300"],
    ["2026-10-02", "food", "n/a"],
    ["2026-10-03", "bills", "4,500"],
    ["2026-10-03", "food", "800"],
    ["2026-10-04", "transport", ""],
    ["2026-10-05", "shopping", "2,100"],
    ["2026-10-06", "food", "650"],
    ["2026-10-06", "bills", "1,200"],
]

with open ("data/expenses.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)

class ExpenseReport:
    """A set of methods to analyze and store results of the given data."""
    
    def __init__(self, path):
        """Input a csv file from its path and clean and summarize the integer column to prepare for analysis."""

        self.rows=[]
        self.skipped=0

        with open (path) as f:
            
            for row in csv.DictReader(f):
                cleaned_amount=funcs.safe_int(row["amount"])

                if cleaned_amount is None:
                    self.skipped+=1
                    continue
                self.rows.append({"category": row["category"], "amount": cleaned_amount})
            
    def total(self): 
        """Return the total value of all amounts."""
        total = sum(r["amount"] for r in self.rows)
        return total

    def by_category(self):
        """Return a dictionary of amounts grouped by category."""
        dic_grouped= {}
        for r in self.rows:
            dic_grouped[r["category"]] = dic_grouped.get(r["category"], 0) + r["amount"]
        return dic_grouped

    def top_category(self):
        """Return the category with the maximum amount."""
        totals= self.by_category()
        if not totals:
            return None
        return max(totals, key=totals.get)
    
    def summary(self):
        """Return a summary of the report analysis"""
        lines = []
        lines.append(f"Categories: {len(self.by_category())}")
        lines.append(f"Total amount: {self.total():.1f}")
        lines.append(f"Skipped rows: {self.skipped}")
        return "\n".join(lines) 
        
    def save_json(self, path):
        """Save the grouped dictionary to a json file"""
        totals= self.by_category()
        with open(path, "w") as f:
            json.dump(totals, f)


report= ExpenseReport("data/expenses.csv")

print(f"Total: {report.total():,}",
      f"By category: {report.by_category()}",
      f"Top: {report.top_category()}",
      f"Skipped: {report.skipped}",
      f"Summary:",
      f"{report.summary()}",
      sep="\n")

report.save_json("data/json_totals.json")


try:
    report = ExpenseReport("data/missing.csv")
    print(report.total())
except FileNotFoundError:
    print(f"Couldn't find the 'missing' file. Check the path.")
```
