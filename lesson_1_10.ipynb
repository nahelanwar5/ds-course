```python
#B1
def save_lines(path, items):
    with open (path, "w") as f:
        for item in items:
            f.write(f"{item}\n")
    
def load_lines(path):
    with open(path) as f:
        lines = [line.strip() for line in f]
    return lines

save_lines("data/fruits.txt", ["apple", "banana", "kiwi"])
print(load_lines("data/fruits.txt"))
print(load_lines("data/fruits.txt") == ["apple", "banana", "kiwi"])   # True
```

```python
#B2
rows = [
    ["name", "dept", "salary"],
    ["Ali", "Sales", "50000"],
    ["Sara", "Tech", "72,000"],
    ["Omar", "Sales", "N/A"],
    ["Hina", "Tech", "68000"],
    ["Zara", "HR", ""],
    ["Bilal", "HR", "45000"],
]

import csv
with open("data/employees.csv", "w", newline="") as f: #newline makes sure csv handles the newlines and extra breaks aren't added in between
    csv.writer(f).writerows(rows)
```

```python
#B3
def safe_int(s):
    try:
        s=s.replace(",", "")
        return int(s)
    except ValueError:
        return None

dic={}

with open("data/employees.csv") as f: #default is reading, a new salary is created here
    for row in csv.DictReader(f):
        salary=safe_int(row["salary"])
        
        if salary is None:
            continue
        else:
            dic[row["dept"]] = dic.get(row["dept"], 0) + salary

dic           

# #B4
with open("data/dept_totals.csv", "w", newline="") as f:
    writer = csv.writer(f)
    header = ["dept", "total"]
    rows = [[dept, total] for dept, total in dic.items()]
    writer.writerows([header] + rows)
```

```python
#B5
import json
with open("data/dept_totals.json", "w") as f:
    json.dump(dic,f)
with open("data/dept_totals.json") as f:
    loaded = json.load(f)
loaded==dic
```

```python
#B6
def safe_read(path):
    try:
        with open(path) as f:
            text = f.read()
            return text
    except FileNotFoundError:          # which error does a missing file raise? (A6)
        return None

print(safe_read('data/a.txt'), safe_read('data/ghost.txt'))
```

```python
#B7

with open ("data/review.txt", "w") as f:
    f.write("the product is great and the delivery was fast\n")
    f.write("the price is fair and the support team is great")
with open ("data/review.txt") as f:
    text= f.read()
    
def count_words(text):
    counts={}
    word_list= text.split()
    for word in word_list:
        counts[word]= counts.get(word, 0)+1
    return counts

def most_common(text):
    counts= count_words(text)
    maxim= max(counts, key=counts.get)
    return maxim

counts=count_words(text)
maxx=most_common(text)
desc_word_list=sorted(counts, key=counts.get, reverse=True)
print(counts, maxx, desc_word_list[:3])
```

```python
def count_lines(path):
    count=0
    try:
        with open (path) as f:
            
            for line in f:
                if not line.strip():     # empty after stripping → skip
                    continue
                else:
                    count +=1
            return count
            
    except FileNotFoundError:
        return None
# lines = [line.strip() for line in f]
# print(lines, len(lines))

print(count_lines("data/review.txt"))

##"w" will always overwrite what's before
```
